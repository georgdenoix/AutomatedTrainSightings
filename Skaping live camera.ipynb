```python
import requests
import re
from bs4 import BeautifulSoup
from datetime import datetime
```

```python
url = "https://www.skaping.com/montreal/grand-port/tour-d-observation"
#url = "https://www.skaping.com/montreal/grand-port/tour-d-observation?archives=MTc3MDMwMDYwMA-YQ"

resp = requests.get(url, timeout=10)
resp.raise_for_status()

html = resp.text
```

```python
soup = BeautifulSoup(html, "html.parser")

scripts = soup.find_all("script")

skaping_script = None
```

```python
for script in scripts:
    if script.string and "ImageMedia(" in script.string:
        skaping_script = script.string
        break

if not skaping_script:
    raise RuntimeError("Skaping ImageMedia script not found")
```

```python
skaping_script
```

```python
url_match = re.search(
    r'new\s+ImageMedia\([^,]+,\s*"([^"]+)"',
    skaping_script
)
```

```python
time_match = re.search(
    r'DateUtil\.dbDateToDate\("([^"]+)"\)',
    skaping_script
)
```

```python
image_url = url_match.group(1) if url_match else None
timestamp = time_match.group(1) if time_match else None
```

```python
# Normalize protocol-relative URL
if image_url and image_url.startswith("//"):
    image_url = "https:" + image_url
```
